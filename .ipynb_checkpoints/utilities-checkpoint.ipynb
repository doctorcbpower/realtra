```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from scipy.integrate import romberg
#import sys
#!{sys.executable} -m pip install h5py
import h5py
from scipy.interpolate import interp1d
```

```python
def scaled_hubble_param(redshift,omegatot,omegam,omegal,omegar):
    return np.sqrt((1.-omegatot)*(1+redshift)**2+omegal+omegam*(1+redshift)**3+omegar*(1+redshift)**4)
#    return np.sqrt(omegal+omegam*(1+redshift)**3)

def scaled_hubble_func(redshift,omegatot,omegam,omegal,omegar):
    return 1./scaled_hubble_param(redshift,omegatot,omegam,omegal,omegar)

def scaled_hubble_func_oneplusz(redshift,omegatot,omegam,omegal,omegar):
    return 1./scaled_hubble_param(redshift,omegatot,omegam,omegal,omegar)/(1+redshift)

def get_comovingdistance(z0,z1,omegatot,omegam,omegal,omegar):
    result,error=quad(scaled_hubble_func,z0,z1,args=(omegatot,omegam,omegal,omegar))
    return result

def get_time(z0,z1,omegatot,omegam,omegal,omegar):
    result,error=quad(scaled_hubble_func_oneplusz,z0,z1,args=(omegatot,omegam,omegal,omegar))
    return result
```

```python
def read_header_data(f,mode):
    offset=0

    if mode=='Snap2':
        offset+=16

    f.seek(offset,os.SEEK_SET)

    skip=np.fromfile(f,dtype=np.int32,count=1) # buffer at start of block
    
    offset+=4
    f.seek(offset,os.SEEK_SET)
    npart=np.fromfile(f,dtype=np.int32,count=6)

    offset+=24
    f.seek(offset,os.SEEK_SET)        
    massarr=np.fromfile(f,dtype=np.float64,count=6)

    offset+=48
    f.seek(offset,os.SEEK_SET)        
    time=np.fromfile(f,dtype=np.float64,count=1)[0]

    offset+=8
    f.seek(offset,os.SEEK_SET)        
    redshift=np.fromfile(f,dtype=np.float64,count=1)[0]

    offset+=8
    f.seek(offset,os.SEEK_SET)        
    flag_sfr=np.fromfile(f,dtype=np.int,count=1)[0]

    offset+=4
    f.seek(offset,os.SEEK_SET)        
    flag_feedback=np.fromfile(f,dtype=np.int32,count=1)[0]

    offset+=4
    f.seek(offset,os.SEEK_SET)        
    nall=np.fromfile(f,dtype=np.int32,count=6)

    offset+=24
    f.seek(offset,os.SEEK_SET)        

    flag_cooling=np.fromfile(f,dtype=np.int32,count=1)[0]
    offset+=4
    f.seek(offset,os.SEEK_SET)        
    num_files=np.fromfile(f,dtype=np.int32,count=1)[0]

    offset+=4
    f.seek(offset,os.SEEK_SET)        

    global BoxSize
    BoxSize=np.fromfile(f,dtype=np.float64,count=1)[0]

    offset+=8
    f.seek(offset,os.SEEK_SET)        

    global Omega0
    Omega0=np.fromfile(f,dtype=np.float64,count=1)[0]              

    offset+=8
    f.seek(offset,os.SEEK_SET)        
    OmegaLambda=np.fromfile(f,dtype=np.float64,count=1)[0]

    offset+=8
    f.seek(offset,os.SEEK_SET)        
    HubbleParam=np.fromfile(f,dtype=np.float64,count=1)[0]

    return npart,massarr,nall,time,redshift,flag_sfr,flag_feedback,num_files,BoxSize,Omega0,OmegaLambda,HubbleParam

def read_position_data(f,mode,ntotal):
#   Handle offset from header
    offset=264
    if mode=='Snap2':
        offset+=16
#   Now handle offset for this block
    offset+=4
    if mode=='Snap2':
        offset+=16
    
    f.seek(offset,os.SEEK_SET)
    pos=np.fromfile(f,dtype=np.dtype(np.float32),count=3*ntotal).reshape(ntotal,3)

    return pos

def read_velocity_data(f,mode,ntotal):
#   Handle offset from header
    offset=264
    if mode=='Snap2':
        offset+=16
#   Handle offset from positions block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16       
#   Now handle offset for this block
    offset+=4
    if mode=='Snap2':
        offset+=16
    
    f.seek(offset,os.SEEK_SET)
    vel=np.fromfile(f,dtype=np.dtype(np.float32),count=3*ntotal).reshape(ntotal,3)
    
    return vel
    
def read_ids_data(f,mode,ntotal):
#   Handle offset from header
    offset=264
    if mode=='Snap2':
        offset+=16
#   Handle offset from positions block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from velocities block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16                
#   Now handle offset for this block
    offset+=4
    if mode=='Snap2':
        offset+=16
    
    f.seek(offset,os.SEEK_SET)

    ids=np.fromfile(f,dtype=(np.uint32),count=ntotal)    

    return ids

def read_masses_data(f,mode,ntotal,massarr):
#   Handle offset from header
    offset=264
    if mode=='Snap2':
        offset+=16
#   Handle offset from positions block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from velocities block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16                
#   Handle offset from IDs block
    offset+=8+ntotal*4
    if mode=='Snap2':
        offset+=16                
#   Now handle offset for this block
    offset+=4
    if mode=='Snap2':
        offset+=16
    
    f.seek(offset,os.SEEK_SET)

    mass=np.fromfile(f,dtype=np.float32,count=np.sum(npart[massarr==0]))

    return masses

def read_internal_energies_data(f,mode,npart,massarr):
    ntotal=np.sum(npart)
#   Handle offset from header
    offset=264
    if mode=='Snap2':
        offset+=16
#   Handle offset from positions block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from velocities block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from IDs block
    offset+=8+ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from masses block
    offset+=8+np.sum(npart[massarr==0])*4
    if mode=='Snap2':
        offset+=16
#   Now handle offset for this block
    offset+=4
    if mode=='Snap2':
        offset+=16
    
    f.seek(offset,os.SEEK_SET)

    u=np.fromfile(f,dtype=np.float32,count=npart[0])
    
    return u

def read_densities_data(f,mode,npart,massarr):
    ntotal=np.sum(npart)    
#   Handle offset from header
    offset=264
    if mode=='Snap2':
        offset+=16
#   Handle offset from positions block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from velocities block
    offset+=8+3*ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from IDs block
    offset+=8+ntotal*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from masses block
    offset+=8+np.sum(npart[massarr==0])*4
    if mode=='Snap2':
        offset+=16
#   Handle offset from internal energies block
    offset+=8+ngas*4
    if mode=='Snap2':
        offset+=16
#   Now handle offset for this block
    offset+=4
    if mode=='Snap2':
        offset+=16
    
    f.seek(offset,os.SEEK_SET)

    rho=np.fromfile(f,dtype=np.float32,count=npart[0])
    
    return rho
        
def write_header_data_in_hdf5(f,npart,massarr,nall,time,redshift,num_files,BoxSize,Omega0,OmegaLambda,HubbleParam):
    hdf5_header=f.create_group('Header')
    hdf5_header.attrs['MassTable']=massarr
    hdf5_header.attrs['NumPart_ThisFile']=npart.astype(np.uint32)
    hdf5_header.attrs['NumPart_Total']=nall.astype(np.uint32)
    hdf5_header.attrs['NumPart_Total_HighWord']=np.zeros(6).astype(np.uint32)
    hdf5_header.attrs['NumFilesPerSnapshot']=num_files
    hdf5_header.attrs['Time']=time
    hdf5_header.attrs['Redshift']=redshift
    hdf5_header.attrs['Omega0']=Omega0
    hdf5_header.attrs['OmegaLambda']=OmegaLambda
    hdf5_header.attrs['HubbleParam']=HubbleParam
    hdf5_header.attrs['BoxSize']=BoxSize
    hdf5_header.attrs['Flag_Entropy_ICs']=0
            
def write_data_in_hdf5(f,pos,vel,ids,nall):
    nstart=nall[0]
    for i in range(6):
        nfinish=nstart+nall[i]
        if nall[i]>0:
            ptype='PartType'+str(i)
            hdf5_ptype=f.create_group(ptype)
            hdf5_ptype.create_dataset('Coordinates',data=pos[nstart:nfinish])
            hdf5_ptype.create_dataset('Velocities',data=vel[nstart:nfinish])
            hdf5_ptype.create_dataset('ParticleIDs',data=ids[nstart:nfinish])
        nstart=nfinish
    
    
```
