```python
import numpy as np
import matplotlib.pyplot as plt
import h5py
```

```python
#define floats and ints
FloatType = np.float64  # double precision: np.float64, for single use np.float32
IntType = np.int32

UnitLengthInCGS=3.0856e21       # 1 kpc
UnitMassInCGS=1.989e43          # 1e10 solar mass
UnitVelocityInCGS=1.e5          # 1 km/s
UnitTimeInCGS=UnitLengthInCGS/UnitVelocityInCGS

k_boltzmann=1.3806e-16            # Boltzmann constant in ergs/K
m_H=1.67e-24                    # Hydrogen mass in g
gamma=FloatType(5./3.)  
delta200=200.
rhocrit=27.755e-9
 
def fm_hernquist(x):
    return x*x/(1+x)**2

def get_enclosed_mass(mnorm,ascale,rin,rout):
    menc_in=mnorm*fm_hernquist(rin/ascale)
    menc_out=mnorm*fm_hernquist(rout/ascale)
    return (menc_out-menc_in)

def get_mass_profile(mnorm,ascale,rin,rout,npts):
    lx=np.linspace(np.log10(rin),np.log10(rout),100)
    mx=get_enclosed_mass(mnorm,ascale,rin,10**lx)
    return lx,mx

def generate_positions(lx,mx,npts):
    fmx=np.array(mx/mx[len(mx)-1])  # Normalise the mass profile to interval [0,1]
    rx=np.random.rand(npts)         # Generate random numbers, for mass shells
    rlx=np.interp(rx,fmx,lx)        # This returns radii
    radius=10**rlx
    x=1.-2.*np.random.rand(npts)    # Now generate random values of x on unit sphere
    rcynd=np.sqrt(1.-x*x)           # Calculate the cylindrical radius
    phi=2.*np.pi*np.random.rand(npts)  # Generate angles uniformly between 0 and 2pi
    y=rcynd*np.cos(phi)             # Now generate y value...
    z=rcynd*np.sin(phi)             # ... and z value
    x*=radius
    y*=radius
    z*=radius
    return x,y,z
```

```python
m200=1
r200=(3*m200/4/np.pi/delta200/rhocrit)**(1./3.)
msigma=2.24e-4
ascale=7.67
bscale=4.0e-3
rin=1.e-10
rout=4.e-3
fgas=0.16
BoxSize=FloatType(2.2*rout)
menc_tot=get_enclosed_mass(m200,ascale,rin,rout)
menc_gas=fgas*get_enclosed_mass(m200,ascale,rin,rout)
menc_sigma=get_enclosed_mass(msigma,bscale,rin,rout)
print("Virial radius: %8.4g kpc"%r200)
print("Virial mass: %8.4g Solar masses"%10**(np.log10(m200)+10))
print("R200/a: %8.4g"%(r200/ascale))
print("Inner radius: %8.4lf kpc (%8.4lf/a)"%(rin,rin/ascale))
print("Outer radius: %8.4lf kpc (%8.4lf/a)"%(rout,rout/ascale))
print("Total enclosed mass: %8.4g Solar masses"%10**(np.log10(menc_tot)+10))
print("Total enclosed gas mass: %8.4g Solar masses"%10**(np.log10(menc_gas)+10))
print("Total enclosed stellar mass: %8.4g Solar masses"%10**(np.log10(menc_sigma)+10))
```

```python
ntab=100
lx,mx=get_mass_profile(m200,ascale,rin,rout,ntab) # Get radius and mass arrays

npts=1000
(x,y,z)=generate_positions(lx,mx,npts)

initial_gas_temperature=1.e4

initial_internal_energy=k_boltzmann*initial_gas_temperature/( gamma - 1.0 )/m_H
initial_internal_energy*=(1./UnitVelocityInCGS**2)

gas_particle_mass=menc_gas/npts

Pos=np.zeros([npts,3],dtype=FloatType)
Pos[:,0]=x
Pos[:,1]=y
Pos[:,2]=z

output_dir='/Users/cpower'   
filename='gas_halo_ics'
IC=h5py.File(output_dir+'/'+filename+'.hdf5', 'w')
## create hdf5 groups
header = IC.create_group("Header")
part0 = IC.create_group("PartType0")
## header entries
NumPart = np.array([npts, 0, 0, 0, 0, 0], dtype = IntType)
header.attrs.create("NumPart_ThisFile", NumPart)
header.attrs.create("NumPart_Total", NumPart)
header.attrs.create("NumPart_Total_HighWord", np.zeros(6, dtype = IntType) )
header.attrs.create("MassTable", np.zeros(6, dtype = IntType) )
header.attrs.create("Time", 0.0)
header.attrs.create("Redshift", 0.0)
header.attrs.create("BoxSize", BoxSize)
header.attrs.create("NumFilesPerSnapshot", 1)
header.attrs.create("Omega0", 0.0)
header.attrs.create("OmegaB", 0.0)
header.attrs.create("OmegaLambda", 0.0)
header.attrs.create("HubbleParam", 1.0)
header.attrs.create("Flag_Sfr", 0)
header.attrs.create("Flag_Cooling", 0)
header.attrs.create("Flag_StellarAge", 0)
header.attrs.create("Flag_Metals", 0)
header.attrs.create("Flag_Feedback", 0)
if Pos.dtype == np.float64:
    header.attrs.create("Flag_DoublePrecision", 1)
else:
    header.attrs.create("Flag_DoublePrecision", 0)

## copy datasets
part0.create_dataset("ParticleIDs", data = np.arange(1, npts+1) )
part0.create_dataset("Coordinates", data = Pos)
part0.create_dataset("Masses", data = gas_particle_mass*np.ones(npts))
part0.create_dataset("Velocities", data = np.zeros(npts))
part0.create_dataset("InternalEnergy", data = initial_internal_energy*np.ones(npts))

## close file
IC.close()
```
